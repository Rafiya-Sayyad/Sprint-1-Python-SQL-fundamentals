### Variables
A variable is a name that we give to a value so that we can use that value later in our program.

#### Example:
A simple real-world example is a labeled box.
If a box has a label called age and we put 22 inside it, we can use the label age whenever we want to access that value.

name = "Rafiya"
age = 22

```python
name = "Rafiya" 
age = 22 
city = "Hyderabad" 
print(name) 
print(age) 
print(city) 
```

### Datatypes

Data types in Python define the type of value stored in a variable and determine the operations that can be performed on that data. 

int – used for whole numbers, such as 22
float – used for decimal numbers, such as 5.4
str – used for text, such as "Rafiya"
bool – used for True or False values

```python
name = "Rafiya" 
age = 22
print(type(name)) 
print(type(age) )
```

### Type Casting

Type casting means changing a value from one data type to another data type.

int() – converts a value to an integer
float() – converts a value to a decimal number
str() – converts a value to a string
bool() – converts a value to a Boolean

#### Example:
we can convert a string into an integer or an integer into a float.

```python
age = "22" 
price = 100 
age = int(age) 
price = float(price) 

print(age) 
print(type(age)) 
print(price) 
print(type(price)) 
```

### if / elif / else

if checks a condition, and if it is True, the code inside the if block runs. 
elif checks another condition if the previous if condition is False.
else runs when all the previous conditions are False.

#### Example:
we can check a student's marks:

If marks are 90 or above → Grade A
If marks are 75 or above → Grade B
Otherwise → Grade C

The program checks the conditions one by one and executes the matching block.

```python
marks = 82 
if marks >= 90: 
    print("Grade A") 
elif marks >= 75: 
    print("Grade B") 
else: 
    print("Grade C")

#Since 82 is not greater than or equal to 90, the if block is skipped.
#The else block is not executed because one condition was already True.
```

### for loop

A for loop is used to repeat a block of code for each item in a sequence.

Example:if we have a list of fruits and want to print each fruit, instead of writing print() separately for every fruit, we can use a for loop.


```python
fruits = ["Apple", "Mango", "Orange"] 
for fruit in fruits: 
    print(fruit)
```

### while loop

A while loop is used to repeat a block of code as long as a condition is True.

Example:
we can print numbers from 1 to 5.
The loop will continue as long as the number is less than or equal to 5.

```python
number = 1 
while number <= 5: 
    print(number) 
    number += 1
```

### break

break is used to completely stop a loop when a particular condition is met.

#### Example:

For example, we want to print numbers from 1 to 10, but we want to stop the loop when the number reaches 5 we use break.

```python
for number in range(1, 11): 
    if number == 5: 
        break 
    print(number)
```

### continue

The continue statement is used to skip the current iteration of a loop.

When Python reaches continue, it skips the remaining code for that iteration and moves to the next iteration.

It is useful when we want to skip a particular value or condition without stopping the entire loop.

#### Example:
we want to print numbers from 1 to 5, but we want to skip the number 3.

```python
for number in range(1, 6): 
    if number == 3: 
        continue 
    print(number)
```

### pass

The pass statement is used when we want to leave a block of code empty.

Python does not allow an empty block after statements like if, for, or a function definition. pass can be used as a placeholder when we plan to write the code later.

#### EXample:
We are creating a program and have an if condition, but we haven't decided what code to write inside it yet.
We can use pass temporarily instead of leaving the block empty.

```python
age = 20 
if age >= 18: 
    pass 
print("Program continues")
```
