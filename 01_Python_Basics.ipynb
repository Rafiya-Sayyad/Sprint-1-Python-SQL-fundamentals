A set is a collection used to store multiple values.

A set does not allow duplicate values. It is also unordered, so we cannot access its items using an index like we do with a list.

Sets are written using curly brackets {}.

Sets are useful when we want to store unique values or perform operations like union, intersection, and difference.

### List Comprehension
#### Comcept:
List comprehension is a shorter way of creating a new list from an existing sequence.

It allows us to write a loop and optionally a condition in a single line of code.

It is useful when we want to create a list in a simple and readable way.

#### Example:
Suppose we want to create a list containing the squares of numbers from 1 to 5. Normally, we would use a for loop:

numbers = [1, 2, 3, 4, 5]
squares = []
for number in numbers:
    squares.append(number ** 2)
print(squares)

With list comprehension, we can write the same logic more shortly.

```python
numbers = [1, 2, 3, 4, 5] 
squares = [number ** 2 for number in numbers] 
print(squares)
```

#### Explanation
- number ** 2 calculates the square of each number.
- for number in numbers takes each number from the list.
- The result of each calculation is stored in the new squares list.

```python
#### List Comprehension with a Condition - We can also use an if condition.

#For example, we can create a list containing only even numbers.

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = [number for number in numbers if number % 2 == 0]
print(even_numbers)
```

#### Explanation
- The loop checks every number in the list.
- number % 2 == 0 checks whether the number is even.
- Only numbers that satisfy the condition are added to the new list.

### Dictionary Comprehension

### Generators
#### Concept

A generator is a special type of function that produces values one at a time instead of returning all values at once. Generators use the yield keyword.

The main advantage of a generator is that it saves memory, especially when working with a large amount of data.

Instead of creating and storing all the values at once, a generator creates each value only when needed.

#### Real world example
Imagine a machine that gives you one item whenever you ask for it.

Instead of producing 100 items and storing them all, it produces:

Item 1 → when needed
Item 2 → when needed
Item 3 → when needed

A generator works in a similar way.

```python
def generate_numbers(): 
    for number in range(1, 6): 
        yield number 
numbers = generate_numbers() 
for number in numbers: 
    print(number)
```

#### Explanation
- generate_numbers() is a generator function.
- yield number produces one number at a time.
- range(1, 6) generates numbers from 1 to 5.
- The for loop gets each value from the generator.
- The generator does not need to store all the numbers in a list.

### yield
#### Concept:

The yield keyword is used inside a generator function to produce a value one at a time.

When yield is reached:
- It gives the value to the caller.
- The function pauses.
- When the next value is requested, the function continues from where it stopped.

#### Example:
Imagine a person giving you books one by one.

They give you:

Book 1 → pause
Book 2 → pause
Book 3 → pause

They don't give you all the books at once.

yield works in a similar way.

```python
def even_numbers(): 
    for number in range(1, 11): 
        if number % 2 == 0: 
            yield number 
numbers = even_numbers() 
for number in numbers: 
    print(number)
```

#### Explanation
- even_numbers() is a generator function.
- The for loop checks numbers from 1 to 10.
- number % 2 == 0 checks whether the number is even.
- yield number gives the even number one at a time.
- The function pauses after each yield and continues when the next value is requested.

### Decorators
#### Concept
A decorator is a function that adds extra functionality to another function without changing the original function's code.
#### why it is used
Decorators are commonly used for:

Logging,
Checking permissions,
Measuring execution time,
Authentication.

#### Example:
Think of a gift wrapper.

The gift is already there, but the wrapper adds something extra without changing the gift itself.

Similarly, a decorator adds extra functionality to a function.

```python
def add_wrapper(func):
    def wrapper():
        print("Gift is being wrapped")
        func()
        print("Gift is ready") 
    return wrapper
@add_wrapper
def gift():
    print("This is the gift")
gift()
```

### nonlocal
#### Definition

The nonlocal keyword is used inside a nested function to modify a variable belonging to the outer function.

#### Real-world example

Imagine a shopping cart. The cart belongs to a particular customer, and an inner function can update the number of items in that customer's cart.

```python
def shopping_cart():
    items = 0
    def add_item():
        nonlocal items
        items += 1
    add_item()
    add_item()
    print("Items in cart:", items)
shopping_cart()
```

#### Explanation
- items belongs to the outer function.
- add_item() is the inner function.
- nonlocal items allows the inner function to modify items.
- Two items are added, so the final count is 2

### map()

map() applies a function to every item in an iterable and returns the results.

### filter()
#### Definition

The filter() function is used to select items from an iterable based on a given condition. It returns a filter object.

#### Real-Life Example

Imagine a teacher wants to find only the students who passed the exam.

Instead of checking and selecting students one by one, the same condition is applied to all students.

```python
marks = [35, 60, 45, 80, 90]
result = list(filter(lambda x: x >= 50, marks))
print(result)
```

- lambda x: x >= 50 selects marks that are 50 or above.
- list() converts the filter object into a list.
- The final result contains only the passing marks.
