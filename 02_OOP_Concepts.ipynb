### Class
#### Concept:
A class is a user-defined blueprint, template, or prototype used to create objects. It defines the structure, characteristics, and behaviors that the resulting objects will have.
A class serves as a logical framework. It does not occupy memory until an actual instance (an object) of that class is created.

It defines:
- What data an object will have → variables/attributes.
- What an object can do → methods/functions.


#### Real world Analogy
Think about a house blueprint.

The blueprint describes:
Number of rooms
Doors
Windows
Other features

The blueprint is not the actual house. It is used to build houses.

Similarly, a Python class is a blueprint, and objects are created using that class.

#### Implementation
##### What does this program do?

This program defines a Student class that stores a student's name and marks.

It also contains a result() method that checks whether the student has passed or failed based on their marks.

Two student objects are created using the same class, demonstrating how one class can be used to create multiple objects.

##### Which part of the code demonstrates the concept?

The class Student: statement demonstrates the concept of a class by defining a blueprint for student objects.

The __init__() method defines the data that each student object will store.

The Student("Aishu", 28) and Student("Rahul", 75) statements create objects using the Student class.

The result() method demonstrates the behavior that the objects can perform.

##### How does the concept work in this implementation?

When Python executes the Student class definition, it creates a blueprint containing the attributes and method defined inside it.

When s1 and s2 are created, Python creates two separate objects using that blueprint.

Each object stores its own name and marks values, while both objects can use the same result() method.

This demonstrates how a class provides a reusable structure for creating multiple objects with their own data.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def result(self):
        if self.marks >= 35:
            print(self.name, "has Passed")
        else:
            print(self.name, "has Failed")
s1 = Student("Aishu", 28)
s2 = Student("Rahul", 75)
s1.result()
s2.result()
```

### Object
#### Concept
An object is an instance of a class.
A class provides the structure or blueprint, while an object is the actual entity created using that class.

For example, if Student is a class, then s1 and s2 can be objects of the Student class. Each object can have its own data while using the structure and methods defined by the class.

A class can be used to create many objects.

#### Real world Analogy
Think about a student registration form.

The form defines what information every student should have, such as:

Name
Age
Marks

The form is similar to a class.

When different students fill out the form, each completed form contains different information.

Student → Class

Rafiya → Object

Rahul → Object

Both Rafiya and Rahul are objects created using the same Student class, but their data is different.

#### Implementation
##### What does this program do?

This program defines a Student class and creates two objects, s1 and s2, from that class.

Each object stores a different student's name and marks. The program then displays the information stored in each object.

##### Which part of the code demonstrates the concept?

The Student class provides the structure for the student objects.

The statements:

s1 = Student("Rafiya", 85)
s2 = Student("Rahul", 72)

create two separate objects from the Student class.

Here, s1 and s2 are objects, while Student is the class.

##### How does the concept work in this implementation?

When s1 is created, Python creates an object containing Rafiya's information.

When s2 is created, Python creates another object containing Rahul's information.

Both objects belong to the same Student class, but they contain different values.

This demonstrates that multiple objects can be created from one class, with each object maintaining its own data.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
s1 = Student("Rafiya", 85)
s2 = Student("Rahul", 72)
s1.display()
print()
s2.display()
```

### Constructor (__init__)
#### concept:
A constructor is a special method that runs automatically when an object is created.

In Python, the constructor is usually written as __init__().

It is mainly used to initialize the data of an object when the object is created.

For example, when we create a Student object, we can use __init__() to immediately assign the student's name and marks.

The constructor helps us avoid assigning values separately after creating every object.

#### Real world Analogy
Imagine registering a new student in a college.

When a student is registered, their basic information such as:

Name
Age
Course
is entered immediately.

Similarly, when we create an object in Python, the __init__() method can automatically initialize the object's information.

For example:

Student("Rafiya", 22, "CSE")

When this object is created, the constructor automatically stores these values in the object.

#### Implementation
##### What does this program do?

This program defines a Student class with a constructor that accepts the student's name, age, and course.

Two student objects are created with different values. The constructor automatically assigns these values to the corresponding attributes of each object.

##### Which part of the code demonstrates the concept?

The following method demonstrates the constructor:

def __init__(self, name, age, course):

The __init__() method is automatically called when a new Student object is created.

The following statements create objects and automatically call the constructor:

s1 = Student("Rafiya", 22, "CSE")
s2 = Student("Sonia", 23, "AI")
###### How does the concept work in this implementation?

When s1 is created, Python automatically calls the __init__() method and passes "Rafiya", 22, and "CSE" as arguments.

The constructor then stores these values in self.name, self.age, and self.course.

The same process happens when s2 is created, but with Sonia's information.

This demonstrates how a constructor automatically initializes an object's data when the object is created.

```python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
s1 = Student("Rafiya", 22, "CSE")
s2 = Student("Sonia", 23, "AI")
s1.display()
print()
s2.display()
```

### Instance Variables
#### Concept:
Instance variables are variables that belong to a particular object.

Each object can have its own values for instance variables. This means that changing the value for one object does not normally change the value of the same variable in another object.

Instance variables are usually created inside the __init__() method using self.

For example, if we create two student objects, each student can have a different name, age, and marks.

#### Real world Analogy
Imagine two students:

Rafiya → 22 years old, 85 marks
Sonia → 23 years old, 78 marks

Both students have the same types of information:

Name
Age
Marks

But their actual values are different.

These individual values are similar to instance variables because they belong to their respective student objects.

Rafiya's object → name = Rafiya, age = 22, marks = 85

Sonia's object  → name = Sonia, age = 23, marks = 78

#### Implementation
##### What does this program do?

This program defines a Student class with three instance variables: name, age, and marks.

Two student objects are created with different values. The program then displays the information stored in each object.

##### Which part of the code demonstrates the concept?

These statements create the instance variables:

self.name = name
self.age = age
self.marks = marks

The self keyword connects these variables to the current object.

The objects s1 and s2 each have their own values for these variables.

##### How does the concept work in this implementation?

When s1 is created, its instance variables store Rafiya's information.

When s2 is created, a separate set of instance variables stores Sonia's information.

Although both objects use the same class, their instance variables contain different values.

This demonstrates that instance variables belong to individual objects and allow each object to maintain its own data.

```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)
s1 = Student("Rafiya", 22, 85)
s2 = Student("Sonia", 23, 78)
s1.display()
print()
s2.display()
```

### Class Variables
#### Concept:
A class variable is a variable that is shared by all objects of a class.

It is defined directly inside the class, outside the methods.

Unlike instance variables, which can have different values for different objects, a class variable usually has the same value for all objects.

#### Real-World Example

Imagine a college where all students belong to the same college.

Each student has their own:

Name
Age

But all students have the same:

College name

So, the student's name and age can be instance variables, while the college name can be a class variable because it is shared by all students.

#### Implementation
##### What does this program do?

This program defines a Student class with a class variable called college. Two objects are created, and both objects access the same college name.

##### Which part of the code demonstrates the concept?

The college variable defined directly inside the class is the class variable.

Both s1 and s2 can access this shared variable.

##### How does the concept work in this implementation?

When the class is created, the college variable is created as a shared variable for the class. Both student objects can access the same value without creating a separate college variable for each object.

This implementation demonstrates how class variables store data that is common to all objects of a class.

```python
class Student:
    college = "ABC College"
    def __init__(self, name):
        self.name = name
s1 = Student("Rafiya")
s2 = Student("Sonia")
print(s1.name, s1.college)
print(s2.name, s2.college)
```

### Static Methods
#### Concept
A static method is a method that belongs to a class but does not depend on any particular object.

It is defined using the @staticmethod decorator.

A static method does not use self or cls because it does not need to access instance or class data.

It is useful for performing a general task related to the class.

#### Real-World Example

Imagine a Student class.

Checking whether a number is a valid mark does not require information about a particular student.

So, we can create a static method called check_marks() to perform this general task.

#### Implementation
##### What does this program do?

This program defines a Student class with a static method that checks whether the given marks are valid.

##### Which part of the code demonstrates the concept?

The @staticmethod decorator and the check_marks() method demonstrate the concept of a static method.

##### How does the concept work in this implementation?

The check_marks() method does not need any information from a particular student object or from the class itself.

It simply receives marks and checks whether they are between 0 and 100.

This implementation demonstrates how a static method can perform a general task related to a class without using object or class data.

```python
class Student:
    @staticmethod
    def check_marks(marks):
        return 0 <= marks <= 100
print(Student.check_marks(85))
print(Student.check_marks(120))
```

### Class Methods
#### Step 1 – Understand the Concept

A class method is a method that belongs to the class rather than a particular object.

It is defined using the @classmethod decorator.

A class method takes cls as its first parameter. The cls parameter refers to the class itself.

Class methods are commonly used when we need to work with class variables or perform an operation that affects the class rather than one specific object.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a college has many students.

Every student belongs to the same college, so the college name can be stored as a class variable.

Suppose the college changes its name. Instead of changing the college name separately for every student object, a class method can be used to update the shared class variable.

This is useful when we want to perform an operation related to the class as a whole rather than a particular student.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines a Student class with a class variable called college. It also defines a class method that changes the college name.

Two student objects use the same class variable, so the updated college name is available to both objects.

##### Which part of the code demonstrates the concept?

The @classmethod decorator and the change_college() method demonstrate the concept of a class method.

The cls parameter refers to the Student class and is used to change the class variable college.

##### How does the concept work in this implementation?

When Student.change_college() is called, Python passes the Student class to the cls parameter.

The method then changes cls.college, which is the class variable shared by all student objects.

This implementation demonstrates how a class method can access and modify data that belongs to the class.

```python
class Student:
    college = "ABC College"
    @classmethod
    def change_college(cls, name):
        cls.college = name
s1 = Student()
s2 = Student()
print(s1.college)
Student.change_college("XYZ College")
print(s2.college)
```

### Properties (@property)
#### Step 1 – Understand the Concept

A property allows us to access a method like an ordinary attribute.

In Python, the @property decorator is used to create a property.

It is useful when we want to control how a value is accessed without changing the way we use it.

For example, instead of calling student.get_name(), we can simply use student.name.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a student has a name stored in their student record.

Normally, we may access the name directly using student.name.

But sometimes we may want to perform some calculation or processing before returning a value.

A property allows us to put that logic inside a method while still accessing it like a normal variable.

For example:

student.name

looks like a normal attribute, but Python actually runs the method marked with @property.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines a Student class with a name attribute and a property called student_name.

The property returns the student's name when it is accessed.

##### Which part of the code demonstrates the concept?

The @property decorator above the student_name() method demonstrates the concept of a property.

The method can be accessed using student.student_name without using parentheses.

##### How does the concept work in this implementation?

When student.student_name is accessed, Python automatically calls the method decorated with @property.

The method returns the value stored in self.name.

This implementation demonstrates how a method can be accessed like an attribute using the @property decorator.

```python
class Student:
    def __init__(self, name):
        self.name = name
    @property
    def student_name(self):
        return self.name
student = Student("Rafiya")
print(student.student_name)
```

### Encapsulation
#### Step 1 – Understand the Concept

Encapsulation means keeping data and the methods that work with that data together inside a class.

It also helps control how the data of an object can be accessed or changed.

In Python, encapsulation is commonly implemented using naming conventions such as _variable and __variable.

A double underscore (__) is used when we want to make an attribute harder to access directly from outside the class.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a bank account.

A bank account has sensitive information such as the account balance. We should not allow anyone to directly change the balance to any value they want.

Instead, the bank provides methods such as:

Deposit money
Withdraw money
Check balance

The data and the methods that control the data are kept together.

Similarly, encapsulation in Python allows us to keep an object's data inside the class and control how that data is accessed or modified.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines a BankAccount class with a private balance variable.

The balance is accessed and modified through methods instead of directly changing the variable from outside the class.

##### Which part of the code demonstrates the concept?

The __balance variable demonstrates encapsulation because it is defined with a double underscore.

The deposit() and get_balance() methods provide controlled ways to modify and access the balance.

##### How does the concept work in this implementation?

The balance is stored inside the BankAccount object and is not intended to be accessed directly from outside the class.

The deposit() method controls how money is added to the balance, while get_balance() allows the current balance to be viewed.

This implementation demonstrates how encapsulation keeps data together with the methods that control it.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance
account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())
```

### Abstraction
#### Step 1 – Understand the Concept

Abstraction means hiding unnecessary implementation details and showing only the important features to the user.

It allows us to focus on what an object does instead of how it does it internally.

In Python, abstraction can be implemented using the abc module and abstract methods.

An abstract method defines what a class should do, but the child class provides the actual implementation.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine using an ATM.

When you withdraw money, you only need to:

Insert your card
Enter your PIN
Enter the amount
Collect the money

You do not need to know how the ATM communicates with the bank server or how it checks the account balance internally.

The ATM hides those internal details and provides only the necessary operations.

Similarly, abstraction hides the internal implementation and exposes only the important functionality.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines an abstract Animal class with a method called sound().

The Dog class inherits from Animal and provides its own implementation of the sound() method.

##### Which part of the code demonstrates the concept?

The Animal class is an abstract class, and sound() is an abstract method.

The Dog class implements the sound() method without exposing any internal details of how the sound is produced.

##### How does the concept work in this implementation?

The abstract class defines what the child class should provide, but it does not define the actual behavior.

The Dog class provides the actual implementation of sound().

This implementation demonstrates how abstraction defines required functionality while leaving the implementation details to the child class.

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
class Dog(Animal):
    def sound(self):
        print("Dog barks")
dog = Dog()
dog.sound()
```

### Inheritance
#### Step 1 – Understand the Concept

Inheritance is a feature of Object-Oriented Programming where one class can use the attributes and methods of another class.

The existing class is called the parent class, and the new class is called the child class.

Inheritance helps us reuse existing code instead of writing the same code again.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine we have a general Person class.

Every person can have common information such as a name and can perform common actions such as introducing themselves.

Now, we want to create a Student class.

A student is also a person, so instead of writing the name and introduction functionality again, the Student class can inherit those features from the Person class.

Person → Parent Class
   ↓
Student → Child Class

The child class can use the features of the parent class and can also have its own additional features.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines a Person class with a name and an introduce() method.

The Student class inherits from the Person class and uses the method defined in the parent class.

##### Which part of the code demonstrates the concept?

The statement class Student(Person): demonstrates inheritance.

It means that Student is inheriting the attributes and methods of the Person class.

##### How does the concept work in this implementation?

When the Student object is created, it uses the constructor inherited from the Person class.

The Student object can also use the introduce() method defined in the parent class.

This implementation demonstrates how inheritance allows a child class to reuse the existing features of a parent class.

```python
class Person:
    def __init__(self, name):
        self.name = name
    def introduce(self):
        print("My name is", self.name)
class Student(Person):
    pass
student = Student("Rafiya")
student.introduce()
```

### Multiple Inheritance
#### Step 1 – Understand the Concept

Multiple inheritance is a type of inheritance where one child class inherits from more than one parent class.

This allows the child class to use the attributes and methods available in multiple parent classes.

For example, a class can inherit features from both a Student class and an Employee class.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a person who is both a student and an employee.

A student may have a method such as study(), while an employee may have a method such as work().

If we create a class called WorkingStudent, it can inherit both behaviors.

The WorkingStudent class can use the features from both parent classes.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines two parent classes, Student and Employee.

The WorkingStudent class inherits from both classes and can use the methods from both parents.

##### Which part of the code demonstrates the concept?

The following statement demonstrates multiple inheritance:

class WorkingStudent(Student, Employee):

The class inherits from both Student and Employee.

##### How does the concept work in this implementation?

The WorkingStudent object can access the study() method from the Student class and the work() method from the Employee class.

This implementation demonstrates how one child class can inherit functionality from multiple parent classes.

```python
class Student:
    def study(self):
        print("Rafiya is studying")
class Employee:
    def work(self):
        print("Rafiya is working")
class WorkingStudent(Student, Employee):
    pass
person = WorkingStudent()
person.study()
person.work()
```

### Multilevel Inheritance
#### Step 1 – Understand the Concept

Multilevel inheritance is a type of inheritance where a class inherits from another class, and then a third class inherits from that child class.

This creates a chain of inheritance.

For example:

Grandparent Class
       ↓
 Parent Class
       ↓
 Child Class

The last child class can access the features inherited through the entire chain.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a simple family relationship.

A Person can have a common feature such as a name.

A Student inherits from Person, so a student can use the features of a person.

Then a CollegeStudent inherits from Student, so a college student can use features from both Student and Person.

Person
  ↓
Student
  ↓
CollegeStudent

This is called multilevel inheritance because inheritance happens through multiple levels.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program creates three classes: Person, Student, and CollegeStudent.

Each class inherits from the class above it. The final CollegeStudent object can use methods from both Person and Student.

##### Which part of the code demonstrates the concept?

The following statements demonstrate multilevel inheritance:

class Student(Person):

and

class CollegeStudent(Student):

Here, Student inherits from Person, and CollegeStudent inherits from Student.

##### How does the concept work in this implementation?

When a CollegeStudent object is created, it can access the introduce() method from Person through Student.

It can also access the study() method directly from Student.

This implementation demonstrates how a child class can inherit features through multiple levels of inheritance.

```python
class Person:
    def introduce(self):
        print("I am a person")
class Student(Person):
    def study(self):
        print("I am studying")
class CollegeStudent(Student):
    pass
student = CollegeStudent()
student.introduce()
student.study()
```

### Hierarchical Inheritance
#### Step 1 – Understand the Concept

Hierarchical inheritance is a type of inheritance where multiple child classes inherit from the same parent class.

The parent class contains common features that can be shared by all the child classes.

Each child class can also have its own additional methods or features.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a Vehicle class that contains a common method called start().

Different types of vehicles, such as Car and Bike, can inherit this common functionality from the Vehicle class.

Both child classes can use the common functionality, while each child class can also have its own specific behavior.

This is called hierarchical inheritance because multiple child classes inherit from one parent class.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines a Vehicle class with a common start() method.

The Car and Bike classes inherit from the Vehicle class and can use the start() method.

##### Which part of the code demonstrates the concept?

The following class definitions demonstrate hierarchical inheritance:

class Car(Vehicle):

and

class Bike(Vehicle):

Both Car and Bike inherit from the same Vehicle class.

##### How does the concept work in this implementation?

The Vehicle class provides the common start() method.

The Car and Bike classes inherit this method, so objects of both classes can use it.

This implementation demonstrates how one parent class can provide common functionality to multiple child classes.

```python
class Vehicle:
    def start(self):
        print("Vehicle started")
class Car(Vehicle):
    pass
class Bike(Vehicle):
    pass
car = Car()
bike = Bike()
car.start()
bike.start()
```

### Polymorphism
#### Step 1 – Understand the Concept

Polymorphism means "many forms."

In Object-Oriented Programming, polymorphism allows the same method or operation to behave differently depending on the object using it.

For example, different animals can have a sound() method, but each animal can produce a different sound.

The same method name can therefore be used with different objects to perform different behaviors.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a sound() action.

A dog makes a barking sound, while a cat makes a meowing sound.

The action is the same — sound() — but the result is different depending on the animal.

This is an example of polymorphism because the same method name can represent different behaviors.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines Dog and Cat classes with the same method called sound().

When the method is called for each object, it produces a different result.

##### Which part of the code demonstrates the concept?

The sound() method in both the Dog and Cat classes demonstrates polymorphism.

Both classes use the same method name but provide different behavior.

##### How does the concept work in this implementation?

When dog.sound() is called, Python executes the sound() method defined in the Dog class.

When cat.sound() is called, Python executes the sound() method defined in the Cat class.

This implementation demonstrates how the same method name can perform different actions depending on the object.

```python
class Dog:
    def sound(self):
        print("Dog barks")
class Cat:
    def sound(self):
        print("Cat meows")
dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
```

### Method Overriding
#### Step 1 – Understand the Concept

Method overriding happens when a child class provides its own version of a method that is already defined in the parent class.

The method in the child class has the same name as the method in the parent class.

This allows the child class to change or customize the behavior inherited from the parent class.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a general Animal class with a sound() method.

The parent class may define a general sound, but different animals make different sounds.

A Dog class can override the sound() method and provide its own behavior.

Similarly, a Cat class can also override the same method with its own behavior.

The child classes are replacing the inherited behavior with their own implementation.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines a Person class with a greet() method.

The Student class inherits from Person and overrides the greet() method with its own implementation.

##### Which part of the code demonstrates the concept?

The greet() method inside the Student class demonstrates method overriding.

The parent class and child class both have a method with the same name.

##### How does the concept work in this implementation?

When student.greet() is called, Python uses the greet() method defined in the Student class instead of the version inherited from Person.

This implementation demonstrates how a child class can provide its own behavior for a method inherited from its parent class.

```python
class Person:

    def greet(self):
        print("Hello, I am a person")

class Student(Person):

    def greet(self):
        print("Hello, I am a student")

student = Student()

student.greet()
```

### Method Overloading (Python Approach)
#### Step 1 – Understand the Concept

Method overloading means having multiple methods with the same name but different parameters.

In languages such as Java and C++, we can define multiple methods with the same name and different numbers or types of parameters.

Python does not support traditional method overloading directly. If we define the same method multiple times in a class, the latest definition replaces the previous one.

Instead, Python commonly uses default arguments or *args to achieve similar behavior.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a calculator that can add numbers.

Sometimes we may want to add two numbers, and sometimes we may want to add three numbers.

In Python, instead of creating multiple add() methods, we can use a default argument.

For example:

add(10, 20)
add(10, 20, 30)

The same method can handle different numbers of arguments.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines an add() method that can add either two numbers or three numbers.

It uses a default value for the third parameter, allowing the same method to work with different numbers of arguments.

##### Which part of the code demonstrates the concept?

The c=0 default parameter demonstrates the Python approach to method overloading.

The method can be called with two or three arguments.

##### How does the concept work in this implementation?

When two numbers are passed, the third parameter automatically gets the value 0.

When three numbers are passed, the third value is used in the calculation.

This implementation demonstrates how Python can achieve overloading-like behavior using default arguments instead of defining multiple methods with the same name.

```python
class Calculator:

    def add(self, a, b, c=0):
        print(a + b + c)
calculator = Calculator()

calculator.add(10, 20)
calculator.add(10, 20, 30)
```

### Composition
#### Step 1 – Understand the Concept

Composition is a relationship where one class contains an object of another class as a part of itself.

It represents a strong "has-a" relationship.

The contained object is created inside the main class, so it is closely connected to that class.

For example, a Car has an Engine. The engine is a part of the car.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Think about a car.

A car has an engine that helps it run.

Instead of putting all engine-related functionality directly inside the Car class, we can create a separate Engine class.

The Car class can then contain an Engine object.

This makes the code easier to organize because each class has its own responsibility.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines an Engine class and a Car class.

The Car class creates and contains an Engine object. The car can then use the method provided by the engine.

##### Which part of the code demonstrates the concept?

The following statement demonstrates composition:

self.engine = Engine()

The Car class creates an Engine object and keeps it as part of the car.

##### How does the concept work in this implementation?

When a Car object is created, an Engine object is also created inside it.

The start() method of the Car class uses the start() method of its engine.

This implementation demonstrates a strong has-a relationship, where a car contains an engine as one of its parts.

```python
class Engine:

    def start(self):
        print("Engine started")

class Car:

    def __init__(self):
        self.engine = Engine()

    def start(self):
        self.engine.start()

car = Car()

car.start()
```

### Aggregation
#### Step 1 – Understand the Concept

Aggregation is a relationship where one class uses an object of another class, but both objects can exist independently.

It represents a "has-a" relationship, but the contained object is not completely dependent on the main object.

For example, a Department has Teacher objects, but a teacher can exist even if the department object is removed.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a college department and a teacher.

A department can have a teacher, but the teacher can exist independently of that particular department.

For example, if a department is removed, the teacher does not necessarily have to be removed.

This represents aggregation because the two objects have a relationship but can exist independently.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines a Teacher class and a Department class.

A teacher object is created separately and then passed to the department object.

##### Which part of the code demonstrates the concept?

The following statement demonstrates aggregation:

self.teacher = teacher

The Department class receives an already existing Teacher object instead of creating the teacher itself.

##### How does the concept work in this implementation?

First, a Teacher object is created independently.

Then, that existing teacher object is passed to the Department object.

Both objects can exist independently, which demonstrates the weaker "has-a" relationship of aggregation.

```python
class Teacher:

    def teach(self):
        print("Teacher is teaching")
class Department:

    def __init__(self, teacher):
        self.teacher = teacher

    def start_class(self):
        self.teacher.teach()
teacher = Teacher()
department = Department(teacher)
department.start_class()
```

### Association
#### Step 1 – Understand the Concept

Association is a relationship between two separate classes where their objects interact with each other.

The objects are independent, but they are connected because one object uses or communicates with another object.

Association represents a general relationship between objects.

For example, a Student can interact with a Teacher by asking questions or attending a class.

#### Step 2 – Demonstrate the Concept
#### Real-World Example

Imagine a student and a teacher.

A student can interact with a teacher during a class.

The student and teacher are separate people and can exist independently.

Their relationship exists because they interact with each other.

Similarly, in Python, objects from different classes can interact without one object owning or creating the other.

#### Step 3 – Implementation Explanation
##### What does this program do?

This program defines Student and Teacher classes.

A student object interacts with a teacher object by calling the teacher's teach() method.

##### Which part of the code demonstrates the concept?

The ask_question() method in the Student class accepts a Teacher object and interacts with it.

The Student class does not create or own the Teacher object.

##### How does the concept work in this implementation?

The Teacher object is created separately from the Student object.

The student then uses the teacher object to interact with it.

Both objects can exist independently, but they have a relationship because they interact with each other.

This implementation demonstrates association between two independent classes.

```python
class Teacher:

    def teach(self):
        print("Teacher is teaching")
class Student:
    def ask_question(self, teacher):
        teacher.teach()
        print("Student is asking a question")
teacher = Teacher()
student = Student()

student.ask_question(teacher)
```

### Magic (Dunder) Methods
#### Step 1 – Understand the Concept

Magic methods, also called dunder methods, are special methods in Python that have double underscores before and after their names.

They allow Python objects to work with built-in operations such as ==, +, len(), and print().

For example, __init__() is a magic method that Python automatically calls when an object is created.

#### Step 2 – Demonstrate the Concept

Imagine two students named Rafiya and Sonia. We want to check whether they have the same name.

Normally, Python does not know how to compare two objects created from our own class. We can define the __eq__() magic method to tell Python how to compare them.

When we use ==, Python automatically uses the __eq__() method.

#### Step 3 – Implement the Concept
##### What does this program do?

This program creates two Student objects and checks whether their names are the same.

##### Which part of the code demonstrates the concept?

The __eq__() method demonstrates the magic method concept. It defines how two Student objects should be compared using ==.

##### How does the concept work in this implementation?

When student1 == student2 is used, Python automatically calls the __eq__() method. The method compares the names of both students and returns True if they are the same.

```python
class Student:

    def __init__(self, name):
        self.name = name
    def __eq__(self, other):
        return self.name == other.name
        
student1 = Student("Rafiya")
student2 = Student("Rafiya")

print(student1 == student2)
```

### super()
#### Step 1 – Understand the Concept

super() is used to access methods or the constructor of a parent class from a child class.

It is commonly used when a child class needs to use the functionality already defined in its parent class.

For example: A Student class can inherit from a Person class. The Student class can use super() to call the Person class constructor.

#### Step 2 – Demonstrate the Concept

Imagine Rafiya is a student. Every student is also a person, so the Person class can store the person's name.

The Student class needs to store the course as well. Instead of writing the name initialization again, it can use super() to call the parent class constructor.

#### Step 3 – Implement the Concept
##### What does this program do?

This program creates a Person parent class and a Student child class. The child class uses super() to call the parent class constructor.

##### Which part of the code demonstrates the concept?

The super().__init__(name) statement demonstrates the use of super().

##### How does the concept work in this implementation?

When a Student object is created, super() calls the __init__() method of the parent Person class. The parent class initializes the name, while the child class initializes the course.

```python
class Person:

    def __init__(self, name):
        self.name = name

class Student(Person):

    def __init__(self, name, course):
        super().__init__(name)
        self.course = course

student = Student("Rafiya", "CSE")

print(student.name)
print(student.course)
```

### __str__()
#### Step 1 – Understand the Concept

When we use print() with an object, Python automatically calls the __str__() method if it is defined.

For example: If we have a Student object, we can use __str__() to display the student's name instead of showing a default object representation.

#### Step 2 – Demonstrate the Concept

### __repr__()
#### Step 1 – Understand the Concept

__repr__() is a special method used to define how an object should be represented as a string, mainly for developers and debugging.

It is useful when we want to see important details about an object clearly.

For example: A Student object can use __repr__() to display the student's name and marks.

#### Step 2 – Demonstrate the Concept

Imagine we have a student named Rafiya who scored 85 marks.

Instead of seeing a default object representation, we can define __repr__() to show useful information about the object.

When we use repr() on the object, Python calls the __repr__() method.

#### Step 3 – Implement the Concept
##### What does this program do?

This program creates a Student object with a name and marks and displays its representation using repr().

##### Which part of the code demonstrates the concept?

The __repr__() method demonstrates the concept. It returns a string containing the student's name and marks.

### __len__()
#### Step 1 – Understand the Concept

__len__() is a special method used to define the length of an object.

When we use the len() function with our own object, Python automatically calls the __len__() method.

For example: If a Student object has three subjects, __len__() can be used to return the number of subjects.

#### Step 2 – Demonstrate the Concept

Imagine Rafiya is studying Python, SQL, and OOP.

We can store these subjects inside a Student object. By defining __len__(), we can use len(student) to find the number of subjects.

#### Step 3 – Implement the Concept
##### What does this program do?

### Singleton Pattern (Bonus)
#### Step 1 – Understand the Concept

The Singleton Pattern is a design pattern that ensures a class has only one object (instance).

If we try to create the object multiple times, the same object is returned instead of creating a new one.

For example: A program may need only one configuration manager. Instead of creating multiple objects, the Singleton Pattern allows only one object to be used.

#### Step 2 – Demonstrate the Concept

Imagine a program has a single Singleton object.

We create two variables, a and b, from the same class. The Singleton Pattern makes sure both variables refer to the same object.

#### Step 3 – Implement the Concept
##### What does this program do?

This program creates a Singleton class and tries to create two objects from it. It checks whether both objects are actually the same object.

##### Which part of the code demonstrates the concept?

The __new__() method demonstrates the Singleton Pattern. It creates an object only when one does not already exist.

##### How does the concept work in this implementation?

When the first object is created, Python creates the instance and stores it in _instance. When another object is created, the existing instance is returned instead of creating a new one.

```python
class Singleton:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

a = Singleton()
b = Singleton()

print(a is b)
```
