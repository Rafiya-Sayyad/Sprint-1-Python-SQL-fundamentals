## Notebook 3 – Exception Handling


### try
#### Definition

The try block is used to write code that may cause an exception.

If an exception occurs inside the try block, Python can handle it using an except block.

##### What does this program do?

This program tries to divide two numbers. The code that may cause an exception is written inside the try block.

```python
try:
    number = 10
    result = number / 2
    print(result)
except:
    print("An error occurred")
```

#### Explanation

The code inside the try block runs normally because 10 / 2 is valid.

If an error occurs inside the try block, Python moves to the except block to handle the error.

Important: try is normally used together with except, else, or finally.

### except
#### Definition

The except block is used to handle an exception that occurs in the try block.

It prevents the program from stopping suddenly when an error occurs.

##### What does this program do?

This program tries to divide a number by zero. Since division by zero causes an error, the except block handles the error and displays a message.

```python
try:
    number = 10
    result = number / 0
    print(result)
except:
    print("Cannot divide by zero")
```

#### Explanation

The statement 10 / 0 causes an error.

Python then skips the remaining code in the try block and executes the except block.

### else
#### Definition

The else block is executed when no exception occurs in the try block.

It is used to write the code that should run when the try block is successful.

##### What does this program do?

This program divides two numbers. If the division is successful, the else block prints the result.

```python
try:
    number = 10
    result = number / 2
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("The result is:", result)
```

#### Explanation

The division 10 / 2 is successful, so no exception occurs.

Because there is no exception, Python executes the else block.

If an exception occurs, the else block is skipped.

```python
try:
    number = 10
    result = number / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("The result is:", result)
```

### finally
#### Definition

The finally block is used to write code that always executes, whether an exception occurs or not.

It is commonly used for cleanup tasks, such as closing a file or releasing a resource.

##### What does this program do?

This program tries to divide a number by zero. An exception occurs, but the finally block still executes.

```python
try:
    number = 10
    result = number / 0
    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Program execution is completed")
```

#### Explanation

The statement 10 / 0 causes a ZeroDivisionError.

The except block handles the exception.

After that, the finally block executes.

The finally block runs whether an exception occurs or not.

### raise
#### Definition

The raise statement is used to manually create and raise an exception when a specific condition occurs.

It allows us to stop the normal flow of the program and show an appropriate error message.

##### What does this program do?

This program checks the age of a person. If the age is less than 18, it manually raises a ValueError.

```python
age = 15

if age < 18:
    raise ValueError("Age must be 18 or above")

print("You are eligible")
```

#### Explanation

The value of age is 15.

Since 15 is less than 18, the raise statement creates a ValueError.

The program stops at the raise statement, so "You are eligible" is not printed.

### assert
#### Definition

The assert statement is used to check whether a condition is true.

If the condition is True, the program continues normally.

If the condition is False, Python raises an AssertionError.

##### What does this program do?

This program checks whether a student's marks are greater than or equal to 35.

If the marks are 35 or more, the assertion is successful.

```python
marks = 75

assert marks >= 35

print("Student passed")
```

#### Explanation

The condition marks >= 35 is True because the marks are 75.

Therefore, the program continues and prints "Student passed".

### Custom Exceptions
#### Definition

A custom exception is a user-defined exception created by a programmer to handle a specific situation in a program.

Custom exceptions are created by creating a class that inherits from Python's Exception class.

##### What does this program do?

This program creates a custom exception called AgeError.

If the age is less than 18, the program raises the custom exception.

```python
class AgeError(Exception):
    pass


age = 15

try:
    if age < 18:
        raise AgeError("Age must be 18 or above")

    print("You are eligible")

except AgeError as error:
    print(error)
```

### Built-in Exceptions

1. ValueError - Occurs when a function receives a value of the correct type but the value is not valid.

2. TypeError - Occurs when an operation or function is used with an inappropriate data type.

3. IndexError - Occurs when we try to access an index that does not exist in a sequence such as a list or tuple.

4. KeyError - Occurs when we try to access a dictionary key that does not exist.

5. AttributeError - Occurs when we try to access an attribute or method that an object does not have.

6. NameError - Occurs when we use a variable or name that has not been defined.

7. FileNotFoundError - Occurs when we try to open or access a file that does not exist.

8. ZeroDivisionError - Occurs when we try to divide a number by zero.

9. ImportError - Occurs when Python cannot import a required name or component from a module.

10. ModuleNotFoundError - Occurs when Python cannot find the module that we are trying to import.

```python
### value error
#### Program that generates the exception

number = int("hello")
print(number)
```

```python
#### Correct way to handle it

try:
    number = int("hello")
    print(number)
except ValueError:
    print("Please enter a valid number")
```

```python
### Type error
#### Program that generates the exception
number = 10
text = "5"

result = number + text
print(result)
```

```python
#### Correct way to handle it
try:
    number = 10
    text = "5"

    result = number + text
    print(result)
except TypeError:
    print("Cannot add an integer and a string")
```

```python
### Index error 
#### Program that generates the exception
numbers = [10, 20, 30]

print(numbers[5])
```

```python
#### Correct way to handle it
try:
    numbers = [10, 20, 30]

    print(numbers[5])
except IndexError:
    print("Index does not exist")
```

```python
### Key error 
#### Program that generates the exception
student = {
    "name": "Riya",
    "age": 22
}

print(student["marks"])
```

```python
#### Correct way to handle it
try:
    student = {
        "name": "Riya",
        "age": 22
    }

    print(student["marks"])
except KeyError:
    print("Key does not exist in the dictionary")
```

```python
### AttributeError
#### Program that generates the exception
name = "Riya"

name.append("Hello")
```

```python
#### Correct way to handle it
try:
    name = "Riya"

    name.append("Hello")
except AttributeError:
    print("This object does not have the requested method")
```

```python
### NameError
#### Program that generates the exception
print(age)
```

```python
#### Correct way to handle it
try:
    print(age)
except NameError:
    print("The variable has not been defined")
```

```python
### FileNotFoundError
#### Program that generates the exception
file = open("student.txt", "r")
```

```python
#### Correct way to handle it
try:
    file = open("student.txt", "r")
    print(file.read())
    file.close()
except FileNotFoundError:
    print("File not found")
```

```python
### ZeroDivisionError
#### Program that generates the exception
number = 10
result = number / 0

print(result)
```

```python
#### Correct way to handle it
try:
    number = 10
    result = number / 0

    print(result)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
### ImportError
#### Program that generates the exception
from math import hello
```

```python
#### Correct way to handle it
try:
    from math import hello
except ImportError:
    print("The required name cannot be imported")
```

```python
### ModuleNotFoundError
#### Program that generates the exception
import mymodule
```

```python
#### Correct way to handle it
try:
    import mymodule
except ModuleNotFoundError:
    print("Module not found")
```
