## Notebook 4 – Logging
### What is Logging?
Logging is the process of recording information about what happens while a program is running.

Python provides a built-in logging module that allows us to record messages such as information, warnings, errors, and other important events.

### Why Use Logging?
Logging is used to record important events, information, warnings, and errors that happen while a program is running.

### Why do we use Logging?

Logging is useful because it helps us:
- Track what the program is doing
- Find and understand errors
- Monitor the program while it is running
- Keep a record of important events
- Debug problems without changing the program

#### Real-Life Example

Imagine an ATM machine.

The ATM may record:

User logged in
Money withdrawal started
₹5,000 withdrawn
Insufficient balance
Transaction failed

These records help the bank understand what happened if there is a problem.

Similarly, logging records important events in a Python program.

### print() vs logging

print() displays messages directly on the screen, while logging records messages with different levels and can send them to the console or a file.

### Log Levels
#### Definition

Log levels are used to indicate the importance or severity of a message recorded by the logging system.

Python provides five commonly used log levels:

DEBUG</br>
INFO</br>
WARNING</br>
ERROR</br>
CRITICAL

#### DEBUG
Gives detailed information for debugging.</br>
When developers need detailed information about the program.

#### INFO	
Gives general information about the program.</br>
When something normal and expected happens.
#### WARNING	
Indicates a possible problem.</br>	
When something unexpected happens but the program can continue.
#### ERROR	
Indicates that an error occurred.</br>	
When an operation fails.
#### CRITICAL	
Indicates a very serious error.</br>	
When the program may not be able to continue.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debugging the program")
logging.info("Program started")
logging.warning("Low memory")
logging.error("File not found")
logging.critical("System failure")
```

### Log Formatters
#### Definition

A log formatter is used to decide how a log message should be displayed or saved.

It can include information such as:

Date and time</br>
Log level</br>
Message</br>
Name of the logger

#### Important Format Fields
Format	Meaning
%(asctime)s	- Date and time</br>
%(levelname)s - Log level</br>
%(message)s - Log message</br>
%(name)s - Logger name</br>
%(filename)s - Name of the Python file

### File Handlers
#### Definition

A File Handler is used to save log messages into a file instead of displaying them only on the screen.

Python provides FileHandler through the logging module.

### Console Handlers
#### Definition

A Console Handler is used to send log messages to the console or terminal screen while the program is running.

Python provides StreamHandler for displaying log messages on the console.

### Rotating Logs
#### Definition

Rotating Logs is a technique used to automatically create a new log file when the current log file reaches a specified size.

This prevents one log file from becoming too large.

Python provides RotatingFileHandler for this purpose.

#### Example:
maxBytes=1000 means the log file can grow up to 1000 bytes.</br>
backupCount=3 means Python keeps up to 3 older log files.

### Logging Best Practices
#### Definition

Logging best practices are recommended methods for creating clear, useful, and manageable log messages in a program.

#### Important Best Practices
1. Use the correct log level

Use the appropriate level for each message.

```python
logging.debug("Checking user details")
logging.info("User logged in")
logging.warning("Low storage")
logging.error("File not found")
logging.critical("System is down")
```

2. Write clear messages

Log messages should clearly explain what happened.


```python
logging.info("User login successful")
```

3. Do not log sensitive information

Never store passwords, credit card numbers, API keys, or other sensitive information in logs.


```python
logging.info("Password: 123456")
```

4. Use logging instead of excessive print()

Use logging when you need to monitor or troubleshoot a real application.

```python
logging.info("Database connection successful")
```

5. Use a useful log format

Include information such as time, level, and message.

```python
"%(asctime)s - %(levelname)s - %(message)s"
```

6. Save important logs to files

File logging allows us to check events later.

```python
logging.FileHandler("app.log")
```

7. Avoid unnecessary logging

Do not create too many logs that make it difficult to find important information.

```python
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.warning("Low storage space")
logging.error("Unable to open file")
```
