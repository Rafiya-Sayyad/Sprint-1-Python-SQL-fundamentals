## Notebook 5 – File Handling

### Reading Text Files

#### Definition

Reading a text file means opening an existing text file and retrieving the data stored inside it.

##### Why is it used?

Reading files is used when we want to access information that has already been saved in a file.

#### Real-Life Example

Imagine you have a notes.txt file containing your daily notes.

If you want to see the notes stored in the file, you need to read the file.

```python
file = open("notes.txt", "r")

content = file.read()

print(content)

file.close()
```

### Writing Text Files
#### Definition

Writing a text file means creating a new text file or adding content to an existing file.

##### Why is it used?

Writing files is used when we want to save information permanently in a file.

#### Real-Life Example

Imagine you want to save a student's name and marks in a text file.

Instead of writing the information manually every time, Python can create a file and store the information in it.

```python
file = open("student.txt", "w")

file.write("Name: Rafiya\n")
file.write("Marks: 90")

file.close()
```

### CSV Files
#### Definition

A CSV (Comma-Separated Values) file is a file used to store data in rows and columns, where values are separated by commas.

##### Why is it used?

CSV files are commonly used to store and exchange tabular data such as student records, employee details, and sales information.

```python
import csv

file = open("students.csv", "r")

reader = csv.reader(file)

for row in reader:
    print(row)

file.close()
```

### JSON Files
#### Definition

JSON (JavaScript Object Notation) is a file format used to store and exchange data in a structured way using key-value pairs.

##### Why is it used?

JSON files are commonly used to store structured data and exchange data between applications, websites, and APIs.

```python
import json

file = open("student.json", "r")

data = json.load(file)

print(data)

file.close()
```

### Pickle Files
#### Definition

A Pickle file is a file used to store Python objects such as lists, dictionaries, tuples, and other Python data in a binary format.

##### Why is it used?

Pickle is used when we want to save a Python object and load it again later without creating the object from the beginning.

#### Real-Life Example

Imagine you have a Python list containing student marks:

marks = [85, 90, 78, 92]

You can save this list into a Pickle file and load it later when you need it.

```python
import pickle

marks = [85, 90, 78, 92]

file = open("marks.pkl", "wb")

pickle.dump(marks, file)

file.close()
```

```python
#### Loading the Pickle File
import pickle

file = open("marks.pkl", "rb")

marks = pickle.load(file)

print(marks)

file.close()
```

### Context Manager (with)
#### Definition

A context manager is used to automatically manage resources such as files. The with statement automatically closes the file after the work is completed.

##### Why is it used?

It is used to safely work with files without manually calling close().

#### Real-Life Example

Imagine borrowing a book from a library.

You use the book and return it when you are finished. Similarly, the with statement opens a file, lets us work with it, and automatically closes it when the work is completed.

```python
with open("notes.txt", "r") as file:
    content = file.read()

print(content)
```

### pathlib
#### Definition

pathlib is a Python module used to work with files and directories using simple and readable path objects.

##### Why is it used?

It is used to create file paths, check whether files exist, read files, write files, and work with directories.

#### Real-Life Example

Imagine you want to check whether a file named notes.txt exists on your computer.

Instead of manually handling the file path, pathlib can do it easily.

```python
from pathlib import Path

file_path = Path("notes.txt")

print(file_path.exists())
```

### os
#### Definition

os is a Python module used to interact with the operating system, such as working with files, folders, paths, and environment information.

##### Why is it used?

It is used to perform operating system tasks directly from a Python program.

#### Real-Life Example

Imagine you want to check the files and folders available in your current location.

Python's os module can show them.

```python
import os

files = os.listdir()

print(files)
```

### shutil
#### Definition

shutil is a Python module used to perform high-level file and directory operations such as copying, moving, and deleting files and folders.

##### Why is it used?

It is used when we want to manage files and directories easily using Python.

#### Real-Life Example

Imagine you have a file called notes.txt and you want to make a copy of it as a backup.

Python's shutil module can copy the file for you.

```python
import shutil

shutil.copy("notes.txt", "notes_backup.txt")

print("File copied successfully")
```

### File Validation
#### Definition

File validation means checking whether a file exists and whether it meets certain conditions before using it.

##### Why is it used?

It is used to make sure that the required file is available and safe to process before performing operations on it.

#### Real-Life Example

Imagine a program needs to read a file called student.txt.

Before reading it, the program can check whether the file exists. If the file does not exist, the program can display an appropriate message instead of producing an error.

```python
from pathlib import Path

file_path = Path("student.txt")

if file_path.is_file():
    print("It is a valid file")
else:
    print("It is not a file")
```

### Directory Operations
#### Definition

Directory operations are actions performed on folders, such as creating, listing, renaming, and deleting directories.

##### Why is it used?

Directory operations are used to organize and manage files and folders using Python.

#### Real-Life Example

Imagine you want to create a folder named students to store student files.

Python can create the folder automatically.

#### Implementation

there are three ways to perform this 
- os.mkdir() creates one directory.
- os.makedirs() creates multiple directories.
- os.listdir() lists files and directories.

```python
#### We can use the os module to create a directory.

import os
os.mkdir("students")
print("Directory created successfully")
```

```python
#### Listing Directories - We can use os.listdir() to see the files and folders in the current directory.

import os
items = os.listdir()
print(items)
```

```python
#### Creating Nested Directories- We can use os.makedirs() to create multiple folders at once.

import os
os.makedirs("college/students")
print("Directories created successfully")
```
