### 2. Tables
#### Definition

A table is a structure inside a database that stores data in the form of rows and columns.

#### Real-Life Example

A student table can contain student ID, name, age, and course.

```python
#### Example
CREATE TABLE students (</br>
    student_id INT,</br>
    name VARCHAR(50),</br>
    age INT,</br>
    course VARCHAR(50)</br>
);
```

##### What does this program do?

### 3. Primary Key
#### Definition

A Primary Key is a column that uniquely identifies each row in a table. It cannot contain duplicate or NULL values.

#### Real-Life Example

A student ID can uniquely identify each student.

```python
#### Example
CREATE TABLE students (</br>
    student_id INT PRIMARY KEY,</br>
    name VARCHAR(50),</br>
    age INT,</br>
    course VARCHAR(50)
);
```

##### What does this program do?

This program makes student_id the primary key, so every student must have a unique student ID.

### Foreign Key
#### Definition

A Foreign Key is a column that creates a connection between two tables by referring to the Primary Key of another table.

#### Real-Life Example

A student's department_id can refer to the ID of a department stored in another table.

```python
#### Example
CREATE TABLE departments (</br>
    department_id INT PRIMARY KEY,</br>
    department_name VARCHAR(50)</br>
);</br>

CREATE TABLE students (</br>
    student_id INT PRIMARY KEY,</br>
    name VARCHAR(50),</br>
    department_id INT,</br>
    FOREIGN KEY (department_id) REFERENCES departments(department_id)</br>
);
```

##### What does this program do?

This program creates two tables and uses department_id as a foreign key to connect students with departments.

### 5. Relationships
#### Definition

A relationship describes how data in one table is connected to data in another table.

Common Types</br>
One-to-One</br>
One-to-Many</br>
Many-to-Many</br>
Real-Life Example

One department can have many students. This is a One-to-Many relationship.

```python
#### Example
CREATE TABLE departments (</br>
    department_id INT PRIMARY KEY,</br>
    department_name VARCHAR(50)</br>
);</br>

CREATE TABLE students (</br>
    student_id INT PRIMARY KEY,</br>
    name VARCHAR(50),</br>
    department_id INT,</br>
    FOREIGN KEY (department_id) REFERENCES departments(department_id)</br>
);
```

##### What does this program do?

This program creates a relationship where one department can be associated with multiple students.

### 6. Constraints
#### Definition

Common Constraints</br>

## Queries

### 7. SELECT
#### Definition

The SELECT statement is used to retrieve data from one or more columns of a table.

#### Real-Life Example

A teacher wants to see the names and ages of all students from the student table.

```python
SELECT name, age
FROM students;
```

This program displays the name and age columns from the students table.

```python
SELECT *
FROM students;
```

This program displays all columns and all rows from the students table.

### 8. WHERE
#### Definition

The WHERE clause is used to filter records and return only the rows that satisfy a specified condition.

#### Real-Life Example

A teacher wants to find students who are older than 20 years.

```python
SELECT *
FROM students
WHERE age > 20;
```

```python
This program displays only the students whose age is greater than 20.
```

```python
SELECT name, course
FROM students
WHERE course = 'Python';
```

This program displays the names and courses of students whose course is Python.

### 9. ORDER BY
#### Definition

The ORDER BY clause is used to sort the result of a query in ascending or descending order based on one or more columns.

```python
SELECT *
FROM students
ORDER BY age ASC;
```

This program displays the students in ascending order of age, from youngest to oldest.

```python
SELECT *
FROM students
ORDER BY age DESC;
```

This program displays the students in descending order of age, from oldest to youngest.

### 10. GROUP BY
#### Definition

The GROUP BY clause is used to group rows that have the same value in one or more columns. It is commonly used with aggregate functions such as COUNT(), SUM(), AVG(), MAX(), and MIN().

```python
SELECT course, COUNT(*) AS student_count
FROM students
GROUP BY course;
```

This program groups students based on their course and counts how many students are in each course.

### 11. HAVING
#### Definition

The HAVING clause is used to filter grouped results after the GROUP BY clause. It is commonly used with aggregate functions such as COUNT(), SUM(), and AVG().

```python
SELECT course, COUNT(*) AS student_count
FROM students
GROUP BY course
HAVING COUNT(*) > 3;
```

This program groups students by course and displays only the courses that have more than 3 students.

### 12. DISTINCT
#### Definition

The DISTINCT keyword is used to remove duplicate values from the query result and display only unique values.

```python
SELECT DISTINCT course
FROM students;
```

This program displays each course only once, even if multiple students are enrolled in the same course.

### 13. LIMIT
#### Definition

The LIMIT clause is used to specify the maximum number of rows that should be returned by a query.

```python
SELECT *
FROM students
LIMIT 5;
```

This program displays only the first 5 rows from the students table.

### 14. INNER JOIN

An INNER JOIN returns only the rows that have matching values in both tables.

```python
SELECT students.name, students.dep_id, teachers.teacher_name
FROM students
INNER JOIN teachers
ON students.dep_id = teachers.dep_id;
```

Only matching records from both the tables are retrieved.

### 15. LEFT JOIN

A LEFT JOIN returns all rows from the left table and the matching rows from the right table. If there is no match, NULL is returned.

```python
SELECT Students.Name, Teachers.Teacher_Name

FROM Students LEFT JOIN Teachers

ON Students.Dep_ID = Teachers.Dep_ID;
```

All records from the Students table are retrieved, along with matching records from the Teachers table.

### 16. RIGHT JOIN
#### Definition

A RIGHT JOIN returns all rows from the right table and the matching rows from the left table. If there is no match, NULL is returned.

```python
SELECT students.name, teachers.dep_id, teachers.teacher_name
FROM students
RIGHT JOIN teachers
ON students.dep_id = teachers.dep_id;
```

All records from the Teachers table are retrieved, along with matching records from the Students table.

### 17. FULL JOIN
#### Definition

A FULL JOIN returns all rows from both tables. Matching rows are combined, and unmatched rows contain NULL.

```python
SELECT students.name, teachers.dep_id, teachers.teacher_name
FROM students
FULL JOIN teachers
ON students.dep_id = teachers.dep_id;
```

All records from both tables are retrieved, and NULL is shown where there is no matching record.

### 18. CROSS JOIN
#### Definition

A CROSS JOIN returns every possible combination of rows from the two tables.


```python
SELECT students.name, teachers.teacher_name
FROM students
CROSS JOIN teachers;
```

Every student is combined with every teacher.

## Advanced SQL

### 19. UNION
#### Definition

The UNION operator is used to combine the results of two or more SELECT queries into a single result. It removes duplicate rows from the final result.

```python
SELECT name
FROM students

UNION

SELECT teacher_name
FROM teachers;
```

This program combines the names from the students table and the teachers table into one result.

### 20. CASE
#### Definition

The CASE expression is used to apply conditions in SQL and return different values based on those conditions.

```python
SELECT name, dep_id,
       CASE
           WHEN dep_id = 101 THEN 'Python Department'
           WHEN dep_id = 102 THEN 'SQL Department'
           ELSE 'Other Department'
       END AS department
FROM students;
```

This program checks each student's dep_id and displays a department name based on the condition.

### 21. Subqueries
#### Definition

A Subquery is a query written inside another SQL query. The inner query provides a result that is used by the outer query.

```python
SELECT name
FROM students
WHERE dep_id = (
    SELECT dep_id
    FROM teachers
    WHERE teacher_name = 'Anil'
);
```

The inner query finds the department ID of teacher Anil. The outer query then finds the student who belongs to that department.

### 22. Common Table Expressions (CTEs)
#### Definition

A Common Table Expression (CTE) is a temporary named result set that is created using the WITH keyword and used within a SQL query.

```python
WITH teacher_departments AS (
    SELECT dep_id
    FROM teachers
)
SELECT name, dep_id
FROM students
WHERE dep_id IN (
    SELECT dep_id
    FROM teacher_departments
);
```

The CTE creates a temporary result called teacher_departments containing the department IDs that have teachers. The main query then uses this result to find the matching students.

### 23. Views

A View is a virtual table created from the result of a SQL query. It does not normally store the data separately; instead, it displays data from the underlying tables.

```python
CREATE VIEW student_teacher AS
SELECT students.name, students.dep_id, teachers.teacher_name
FROM students
INNER JOIN teachers
ON students.dep_id = teachers.dep_id;
```

This program creates a view called student_teacher that contains students and their matching teachers.

```python
#### Using the View
SELECT *
FROM student_teacher;
```

```python
#### Deleting a View
DROP VIEW student_teacher;
```

### 25. Window Functions
#### Definition

A Window Function performs a calculation across a set of related rows without combining those rows into a single result.

Unlike GROUP BY, a window function keeps each individual row in the result.

```python
SELECT name, dep_id,
       ROW_NUMBER() OVER (ORDER BY dep_id) AS row_number
FROM students;
```

This program assigns a unique number to each student based on the order of their dep_id, while keeping all students in the result.

```python
#### Important Point

## The basic structure of a window function is:

FUNCTION() OVER (ORDER BY column)
```

### 26. ROW_NUMBER()
#### Definition

The ROW_NUMBER() window function assigns a unique sequential number to each row in the result.

```python
SELECT name, dep_id,
       ROW_NUMBER() OVER (ORDER BY dep_id) AS row_number
FROM students;
```

This program assigns a unique number to each student based on the ascending order of dep_id.

#### Important Point

ROW_NUMBER() always gives a different number to each row, even when two rows have the same value being ordered.

### 27. RANK()
#### Definition

The RANK() window function assigns a rank to each row based on a specified order. If two rows have the same value, they receive the same rank, and the next rank is skipped.

```python
SELECT name, marks,
       RANK() OVER (ORDER BY marks DESC) AS student_rank
FROM students;
```

| name   | marks | student_rank |
| ------ | ----: | -----------: |
| Rafiya |    90 |            1 |
| Sonia  |    85 |            2 |
| Divya  |    85 |            2 |


This program ranks students from the highest marks to the lowest marks.

### 28. DENSE_RANK()
#### Definition

The DENSE_RANK() window function assigns a rank to each row based on a specified order. If two rows have the same value, they receive the same rank, but the next rank is not skipped.

```python
SELECT name, marks,
       DENSE_RANK() OVER (ORDER BY marks DESC) AS student_rank
FROM students;
```

| name   | marks | student_rank |
| ------ | ----: | -----------: |
| Rafiya |    90 |            1 |
| Sonia  |    85 |            2 |
| Divya  |    85 |            2 |


This program ranks students from the highest marks to the lowest marks. Students with the same marks receive the same rank.

#### Important Point

DENSE_RANK() gives the same rank to tied values but does not skip the next rank.

### 29. LAG()
#### Definition

The LAG() window function is used to access the value from a previous row without using a self-join.

```python
SELECT name, marks,
       LAG(marks) OVER (ORDER BY marks DESC) AS previous_marks
FROM students;
```

| name   | marks | previous_marks |
| ------ | ----: | -------------: |
| Rafiya |    90 |           NULL |
| Sonia  |    85 |             90 |
| Divya  |    80 |             85 |


This program displays each student's marks along with the marks of the previous student in the specified order.

### 30. LEAD()
#### Definition

The LEAD() window function is used to access the value from the next row without using a self-join.

```python
SELECT name, marks,
       LEAD(marks) OVER (ORDER BY marks DESC) AS next_marks
FROM students;
```

| name   | marks | next_marks |
| ------ | ----: | ---------: |
| Rafiya |    90 |         85 |
| Sonia  |    85 |         80 |
| Divya  |    80 |       NULL |


This program displays each student's marks along with the marks of the next student in the specified order.
